```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
churn_data = pd.read_csv("churn_data.csv")
print("Shape of the dataset:",churn_data.shape)
churn_data.head()
```

```python
X= churn_data.drop(columns="churn")
y = churn_data["churn"]

numeric_features = ["tenure_months","monthly_charges","support_calls"]
categorical_features = ["contract","internet_service","payment_method"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
print("Taining shape: ", X_train.shape)
print("Testing shape: ", X_test.shape)
print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier
 
numeric_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
 
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
 
preprocessor = ColumnTransformer(transformers=[
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
 
knn_pipeline = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("model", KNeighborsClassifier(n_neighbors=7, weights="distance"))
])
 
knn_pipeline
```

```python
knn_pipeline.fit(X_train,y_train)
print("Preprocessing and KNN model training completed")
```
