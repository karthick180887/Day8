```python
import pandas as pd
import joblib
from pathlib import Path
```

```python
model_path = Path("diabetes_model.joblib")
artifact = joblib.load(model_path)
pipeline = artifact["pipeline"]
features = artifact["features"]
print("model loaded sucessfully")
print("Version: ", artifact.get("version"))
print("Expected Features", features)
```

```python
patients = [
    {
        "Pregnancies": 2, "Glucose": 120, "BloodPressure": 70,
        "SkinThickness": 25, "Insulin": 80, "BMI": 28.5,
        "DiabetesPedigreeFunction": 0.35, "Age": 30
    },
    {
        "Pregnancies": 6, "Glucose": 165, "BloodPressure": 78,
        "SkinThickness": 30, "Insulin": 150, "BMI": 34.2,
        "DiabetesPedigreeFunction": 0.65, "Age": 50
    }
]
 
print("Number of patients:", len(patients))
```

```python
#convert list into dataframe
```

```python
patient_df = pd.DataFrame(patients)
 
missing_features = [f for f in features if f not in patient_df.columns]

if missing_features:

    raise ValueError(f"Missing required features: {missing_features}")
 
patient_df = patient_df[features]

display(patient_df)
 
```

```python
predictions = pipeline.predict(patient_df)
probabilities = pipeline.predict_proba(patient_df)[:, 1]
 
results = patient_df.copy()
results['prediction'] = predictions
results['prediction_label'] = [
    "Diabetes" if p == 1 else "No Diabetes" for p in predictions
]
results['diabetes_probability_percent'] = (probabilities * 100).round(2)
 
display(results)
```
